```python
def kalkulator(num1,operator,num2):
    if operator == "+":
        result = num1 + num2
    elif operator == "-":
        result = num1 - num2
    elif operator == "*":
        result = num1 * num2
    elif operator == "/":
        if num2 != 0:
            result = num1 / num2
        else:
            print("Error: Tidak bisa dibagi dengan nol!")
            return
    print("Hasil: ", result)

kalkulator(1, "*", 3)
```
