### Anomali / Insight pada `master_pizza_type`

Setelah meninjau laporan profil untuk `master_pizza_type` dan melihat sampel data, berikut adalah anomali yang ditemukan:

1.  **Inkonsistensi Huruf Besar/Kecil pada Kolom `category`:**
    *   Beberapa nilai dalam kolom `category` menggunakan huruf kapital (misalnya, 'CHICKEN', 'CLASSIC'), sementara yang lain menggunakan huruf kecil (misalnya, 'veggie').
    *   **Dampak:** Inkonsistensi ini dapat menyebabkan masalah saat melakukan filter, pengelompokan, atau analisis kategorikal. Sistem mungkin memperlakukan 'veggie' dan 'VEGGIE' sebagai dua kategori yang berbeda padahal seharusnya sama.
    *   **Rekomendasi:** Lakukan standarisasi kapitalisasi (misalnya, konversi semua ke huruf kapital atau semua ke huruf kecil) untuk memastikan konsistensi data.

2.  **Nilai `created_date` yang Seragam (2024-12-31 00:00:00):**
    *   Semua entri dalam kolom `created_date` memiliki nilai yang sama persis (`2024-12-31 00:00:00`). Ini sangat tidak biasa untuk timestamp pembuatan record yang sebenarnya.
    *   **Dampak:** Ini mengindikasikan bahwa kolom ini kemungkinan diisi dengan nilai default, placeholder, atau ada kesalahan dalam proses injeksi data, bukan mencerminkan waktu pembuatan data yang sebenarnya. Informasi waktu pembuatan yang akurat tidak tersedia.
    *   **Rekomendasi:** Periksa proses *data ingestion* atau *ETL* untuk memahami mengapa kolom ini memiliki nilai yang seragam. Jika diperlukan, pertimbangkan untuk menangani kolom ini sebagai indikator status atau mengabaikannya untuk analisis berbasis waktu, kecuali jika ada penjelasan yang valid untuk keseragaman ini.

### Anomali / Insight pada `transaction_order_detail`

Setelah meninjau laporan profil untuk `transaction_order_detail`, berikut adalah beberapa poin penting mengenai kualitas data:

1.  **`order_details_id`**: Kolom ini berfungsi sebagai primary key untuk detail pesanan. Pastikan semua nilai unik dan tidak ada nilai yang hilang. Jika ada duplikasi atau nilai kosong, ini mengindikasikan masalah integritas data.

2.  **`order_id`**: Ini adalah foreign key yang menghubungkan ke tabel `transaction_order`. Penting untuk memeriksa:
    *   **Nilai Hilang**: Apakah ada `order_id` yang kosong? Ini berarti ada detail pesanan yang tidak terhubung dengan pesanan induk.
    *   **Integritas Referensial**: Pastikan semua `order_id` yang ada di tabel ini juga ada di tabel `transaction_order`. Jika tidak, berarti ada 'detail pesanan yatim piatu' (orphaned order details).

3.  **`pizza_id`**: Kolom ini adalah foreign key yang menghubungkan ke tabel `master_pizza`. Sama seperti `order_id`:
    *   **Nilai Hilang**: Apakah ada `pizza_id` yang kosong? Ini berarti ada detail pesanan yang tidak merujuk ke jenis pizza yang valid.
    *   **Integritas Referensial**: Pastikan semua `pizza_id` yang ada di tabel ini juga ada di tabel `master_pizza`.

4.  **`quantity`**: Kolom ini menunjukkan jumlah pizza dalam satu detail pesanan.
    *   **Rentang Nilai**: Periksa apakah nilai kuantitas selalu positif dan masuk akal (misalnya, tidak ada 0 atau nilai negatif, dan tidak ada angka yang terlalu besar yang bisa menjadi outlier).
    *   **Distribusi**: Perhatikan distribusi kuantitas. Apakah sebagian besar pesanan hanya satu pizza, atau ada pesanan dengan kuantitas besar?

5.  **`created_date`**: Kolom timestamp ini menunjukkan kapan detail pesanan dibuat.
    *   **Rentang Tanggal**: Periksa rentang tanggal untuk memastikan semua data berada dalam periode waktu yang diharapkan. Adakah tanggal yang di masa depan atau terlalu jauh di masa lalu?
    *   **Konsistensi**: Periksa format tanggal dan waktu. Jika semua `created_date` memiliki nilai yang sama (seperti yang terlihat pada `master_pizza_type`), ini mungkin menunjukkan penggunaan nilai default atau masalah *data ingestion*.

# **ERD**

Data biasanya disimpan di berbagai tabel atau sistem yang berbeda pada RDBMS (Relational Database Management System). Untuk membantu memahami struktur dan hubungan data sebelum melakukan analisis SQL, biasa digunakan ERD (Entity Relationship Diagram).

**ERD** adalah diagram yang digunakan untuk menggambarkan struktur hubungan antar data dalam sebuah database. Langkah-langkah pembuatan ERD adalah sebagai berikut :

* Untuk membuat ERD pada data Pizza Transaction dengan bantuan AI, buka situs https://www.eraser.io/ai/erd-generator

* Ketikan prompt: *"Sebuah table bernama trx_order_detail memiliki kolom 'order_details_id', 'order_id', 'pizza_id', 'quantity', 'created_date'. Table dengan nama trx_order memiliki kolom 'order_id', 'customer_id', 'order_maker_id', 'order_date', 'order_time', 'completion_time', 'is_complain', 'complain_detail', 'created_date'. Kemudian mst_pizza_type memiliki kolom 'pizza_type_id', 'name', 'category', 'ingredients', 'created_date', 'updated_date' lalu tabel mst_pizza memiliki kolom 'pizza_id', 'pizza_type_id', 'size', 'price', 'production_cost', 'created_date', 'updated_date' dan table mst_customer memiliki kolom 'customer_id', 'customer_name', 'gender', 'birth_date', 'phone_number', 'email', 'open_date', 'created_date', 'updated_date'. Buatkan ERD nya!"*
