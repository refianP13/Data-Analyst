# <b>B. <span style='color:#0B2F9F'><code>JOIN</code></span></b>

Data biasanya disimpan di berbagai tabel atau sistem yang berbeda pada RDBMS (Relational Database Management System). JOIN memungkinkan pengguna untuk menggabungkan data dari berbagai sumber untuk mendapatkan gambaran yang lebih lengkap. Memahami query JOIN memungkinkan seorang pengguna untuk mengintegrasikan dan menganalisis data secara efektif, sehingga dapat meningkatkan kualitas analisis dan mendukung pengambilan keputusan berbasis data.

Operasi JOIN standar dapat dilakukan jika terdapat kunci pada tiap tabel yang saling terhubung. Operasi join dasar yang paling umum digunakan ada 4 yakni :
<ul>
    <li><b>LEFT JOIN</b></li>
    LEFT JOIN adalah jenis operasi penggabungan baris dari dua tabel, tetapi hanya mengembalikan semua baris dari tabel kiri (tabel pertama) dan baris yang cocok dari tabel kanan (tabel kedua). Jika tidak ada kecocokan di tabel kanan, maka nilai-nilai untuk kolom dari tabel kanan akan diisi dengan NULL.<br><br>
    <img src="https://learn.microsoft.com/en-us/power-query/media/merge-queries-left-outer/left-outer-join-operation.png" width="40%">
    <br><br>
    <li><b>RIGHT JOIN</b></li>
    RIGHT JOIN adalah jenis operasi penggabungan baris dari dua tabel, tetapi hanya mengembalikan semua baris dari tabel kanan (tabel kedua) dan baris yang cocok dari tabel kiri (tabel pertama). Jika tidak ada kecocokan di tabel kiri, maka nilai-nilai untuk kolom dari tabel kiri akan diisi dengan NULL.<br><br>
    <img src="https://learn.microsoft.com/en-us/power-query/media/merge-queries-right-outer/right-outer-join-operation.png" width="40%">
    <br><br>
    <li><b>INNER JOIN</b></li>
    INNER JOIN adalah jenis operasi penggabungan yang hanya mengembalikan baris yang memiliki kecocokan di kedua tabel berdasarkan kriteria tertentu.<br><br>
    <img src="https://learn.microsoft.com/en-us/power-query/media/merge-queries-inner/inner-join-operation.png" width="40%">
    <br><br>
     <li><b>FULL JOIN</b></li>
    FULL JOIN adalah jenis operasi penggabungan semua baris dari kedua tabel, baik yang memiliki kecocokan maupun yang tidak. Dengan demikian, hasilnya akan mencakup semua data dari tabel kiri dan tabel kanan.<br><br>
    <img src="https://learn.microsoft.com/en-us/power-query/media/merge-queries-full-outer/full-outer-join-operation.png" width="40%">
    <br><br>
</ul>

#### <i><b><span style='color:#55679C'>Quest : </span>Untuk mengetahui informasi mengenai banyak transaksi tiap customer (yang pernah bertransaksi saja); tampilkan customer_id, nama, gender, banyak transaksi yang pernah dilakukan urutkan berdasarkan transaksi terbanyak!

```python
%%bigquery --project {project_id} --verbose

SELECT c.customer_id, c.customer_name, c.gender, COUNT(DISTINCT o.order_id) AS total_order
FROM `dqpizza.master_customer` AS c
INNER JOIN `dqpizza.transaction_order` AS o ON c.customer_id = o.customer_id
GROUP BY c.customer_id, c.customer_name, c.gender
ORDER BY total_order DESC
```

#### <i><b><span style='color:#55679C'>Quest : </span>Kita perlu approch customer terdaftar namun belum pernah bertransaksi (belum pernah membeli pizza) di DQPizza, tampilkan identitas lengkap customer sesuai kriteria tersebut!

```python
%%bigquery --project {project_id} --verbose

SELECT c.customer_id, c.customer_name, c.gender, c.birth_date, c.phone_number, c.email
FROM `dqpizza.master_customer` AS c
LEFT JOIN `dqpizza.transaction_order` AS o ON c.customer_id = o.customer_id
WHERE o.order_id IS NULL
```

#### <i><b><span style='color:#55679C'>Quest : </span>Kategori pizza apa yang menyumbang pendapatan paling besar di DQPizza pada tahun 2024?

```python
%%bigquery --project {project_id} --verbose

SELECT DISTINCT
  UPPER(mpt.category) AS pizza_category,
  COUNT(DISTINCT t.order_id) AS total_order,
  SUM(td.quantity) AS total_quantity,
  SUM(SAFE_CAST(REPLACE(mp.price, 'IDR', '') AS FLOAT64) * td.quantity) AS total_revenue
FROM dqpizza.transaction_order AS t
INNER JOIN dqpizza.transaction_order_detail AS td ON t.order_id = td.order_id
INNER JOIN dqpizza.master_pizza AS mp ON td.pizza_id = mp.pizza_id
INNER JOIN dqpizza.master_pizza_type AS mpt ON mp.pizza_type_id = mpt.pizza_type_id
WHERE EXTRACT(YEAR FROM PARSE_DATE('%m/%d/%Y', t.order_date)) = 2024
GROUP BY UPPER(mpt.category)
ORDER BY total_revenue DESC
```

## **Bagian 1** - `transaction_order`

Ambil kolom *order_id, customer_id, order_maker_id, order_date, order_time, completion_time, duration_in_minutes, is_complain* dan *complain detail* pada table **transaction_order** dengan ketentuan :
1. Pastikan kolom order_date memiliki format 'YYYY-MM-DD', bertipe data DATE dan pastikan nama kolom aliasnya tetap order_date
2. Pastikan kolom order_time dan completion_time memiliki format 'HH:MM:SS', bertipe data TIME lalu pastikan nama kolom aliasnya tetap *order_time* dan *completion_time*
3. Untuk kolom baru *duration_in_minutes* merupakan kolom yang menyatakan lamanya pizza tersaji dari saat pesanan diinput ke sistem dengan kata lain selisih dalam satuan menit antara completion_time dengan order_time
4. Ambil hanya data yang 'ID Pesanannya' nya tidak kosong untuk menjaga validitas transaksi
5. Tampilkan hasilnya sesuai expected output berikut<br><br><img src="https://raw.githubusercontent.com/bachtiyarwork/Image/main/DQPizza%20-%20Project%20Bagian%201.png" width="80%">


```python
%%bigquery --project {project_id} --verbose

SELECT
  order_id,
  customer_id,
  order_maker_id,
  PARSE_DATE('%m/%d/%Y', order_date) AS order_date,
  PARSE_TIME('%H:%M:%S', order_time) AS order_time,
  PARSE_TIME('%H:%M:%S', completion_time) AS completion_time,
  TIME_DIFF(PARSE_TIME('%H:%M:%S', completion_time), PARSE_TIME('%H:%M:%S', order_time), MINUTE) AS duration_in_minutes,
  is_complain,
  complain_detail
FROM `dqpizza.transaction_order`
WHERE order_id IS NOT NULL
```

4. Karena tabel master_customer adalah berjenis tabel master maka pastikan semua nilainya unik setiap barisnya

## **Proses Integrasi**       
Untuk proses integrasi ini akan membutuhkan CTE dalam prosesnya, lengkapi query dibawah ini sesuai petunjuk berikutnya.

1. Pada CTE trx_order isikan query pada **Bagian 1** yang sudah dikonstruksikan sebelumnya
2. Pada CTE trx_order_detail isikan query pada **Bagian 2** yang sudah dikonstruksikan sebelumnya
3. Pada CTE mst_pizza isikan query pada **Bagian 3** yang sudah dikonstruksikan sebelumnya
4. Pada CTE mst_pizza_type isikan query pada **Bagian 4** yang sudah dikonstruksikan sebelumnya
5. Pada CTE mst_customer isikan query pada **Bagian 5** yang sudah dikonstruksikan sebelumnya
6. Relasikan semua CTE-nya (trx_order, trx_order_detail, mst_pizza, mst_pizza_type dan mst_customer) dengan kunci, penggunaan prefix dan metode join yang tepat
7. Tampilkan kolom
    * order_id,
    * order_details_id,
    * pizza_id,
    * customer_id,
    * order_maker_id,
    * order_date,
    * customer_name,
    * customer_gender,
    * customer_age,
    * customer_tenure_in_months,
    * pizza_name,
    * pizza_category,
    * pizza_size,
    * pizza_price,
    * pizza_production_cost,
    * quantity,
    * pizza_ingredients,
    * order_time,
    * completion_time,
    * duration_in_minutes,
    * is_complain,
    * complain_detail

```python
%%bigquery --project {project_id} --verbose

WITH trx_order AS (
    SELECT
      order_id,
      customer_id,
      order_maker_id,
      PARSE_DATE('%m/%d/%Y', order_date) AS order_date,
      PARSE_TIME('%H:%M:%S', order_time) AS order_time,
      PARSE_TIME('%H:%M:%S', completion_time) AS completion_time,
      TIME_DIFF(PARSE_TIME('%H:%M:%S', completion_time), PARSE_TIME('%H:%M:%S', order_time), MINUTE) AS duration_in_minutes,
      is_complain,
      complain_detail
    FROM `dqpizza.transaction_order`
    WHERE order_id IS NOT NULL
),
trx_order_detail AS (
    SELECT
      order_details_id,
      order_id,
      pizza_id,
      quantity
    FROM `dqpizza.transaction_order_detail`
    WHERE order_id IS NOT NULL AND order_details_id IS NOT NULL
),
mst_pizza AS (
    SELECT
      pizza_id,
      pizza_type_id,
      UPPER(size) AS pizza_size,
      CAST(REPLACE(price, 'IDR', '') AS FLOAT64) AS pizza_price,
      CAST(REPLACE(production_cost, 'IDR', '') AS FLOAT64) AS pizza_production_cost
    FROM `dqpizza.master_pizza`
    WHERE pizza_id IS NOT NULL
),
mst_pizza_type AS (
    SELECT
      pizza_type_id,
      UPPER(name) AS pizza_name,
      UPPER(category) AS pizza_category,
      ingredients AS pizza_ingredients
    FROM `dqpizza.master_pizza_type`
    WHERE pizza_type_id IS NOT NULL
    GROUP BY ALL
),
mst_customer AS (
  SELECT DISTINCT
      customer_id,
      customer_name,
      gender AS customer_gender,
      DATE_DIFF(DATE '2025-12-31', COALESCE(SAFE.PARSE_DATE('%Y-%m-%d', birth_date), SAFE.PARSE_DATE('%m/%d/%Y', birth_date), SAFE.PARSE_DATE('%Y/%m/%d', birth_date)), YEAR) AS customer_age,
      DATE_DIFF(DATE '2025-12-31', COALESCE(SAFE.PARSE_DATE('%Y-%m-%d', open_date), SAFE.PARSE_DATE('%m/%d/%Y', open_date), SAFE.PARSE_DATE('%Y/%m/%d', open_date)), MONTH) AS customer_tenure_in_months
  FROM
      dqpizza.master_customer
  WHERE
      customer_id IS NOT NULL
)
SELECT
    o.order_id,
    od.order_details_id,
    od.pizza_id,
    o.customer_id,
    o.order_maker_id,
    o.order_date,
    c.customer_name,
    c.customer_gender,
    c.customer_age,
    c.customer_tenure_in_months,
    pt.pizza_name,
    pt.pizza_category,
    p.pizza_size,
    p.pizza_price,
    p.pizza_production_cost,
    od.quantity,
    pt.pizza_ingredients,
    o.order_time,
    o.completion_time,
    o.duration_in_minutes,
    o.is_complain,
    o.complain_detail
FROM trx_order AS o
INNER JOIN trx_order_detail AS od ON o.order_id = od.order_id
INNER JOIN mst_pizza AS p ON od.pizza_id = p.pizza_id
INNER JOIN mst_pizza_type AS pt ON p.pizza_type_id = pt.pizza_type_id
INNER JOIN mst_customer AS c ON o.customer_id = c.customer_id;
```

```python
%%bigquery --project {project_id} --verbose

CREATE TABLE dqpizza.tbl_detail_transaction AS

WITH trx_order AS (
    SELECT
      order_id,
      customer_id,
      order_maker_id,
      PARSE_DATE('%m/%d/%Y', order_date) AS order_date,
      PARSE_TIME('%H:%M:%S', order_time) AS order_time,
      PARSE_TIME('%H:%M:%S', completion_time) AS completion_time,
      TIME_DIFF(PARSE_TIME('%H:%M:%S', completion_time), PARSE_TIME('%H:%M:%S', order_time), MINUTE) AS duration_in_minutes,
      is_complain,
      complain_detail
    FROM `dqpizza.transaction_order`
    WHERE order_id IS NOT NULL
),
trx_order_detail AS (
    SELECT
      order_details_id,
      order_id,
      pizza_id,
      quantity
    FROM `dqpizza.transaction_order_detail`
    WHERE order_id IS NOT NULL AND order_details_id IS NOT NULL
),
mst_pizza AS (
    SELECT
      pizza_id,
      pizza_type_id,
      UPPER(size) AS pizza_size,
      CAST(REPLACE(price, 'IDR', '') AS FLOAT64) AS pizza_price,
      CAST(REPLACE(production_cost, 'IDR', '') AS FLOAT64) AS pizza_production_cost
    FROM `dqpizza.master_pizza`
    WHERE pizza_id IS NOT NULL
),
mst_pizza_type AS (
    SELECT
      pizza_type_id,
      UPPER(name) AS pizza_name,
      UPPER(category) AS pizza_category,
      ingredients AS pizza_ingredients
    FROM `dqpizza.master_pizza_type`
    WHERE pizza_type_id IS NOT NULL
    GROUP BY ALL
),
mst_customer AS (
  SELECT DISTINCT
      customer_id,
      customer_name,
      gender AS customer_gender,
      DATE_DIFF(DATE '2025-12-31', COALESCE(SAFE.PARSE_DATE('%Y-%m-%d', birth_date), SAFE.PARSE_DATE('%m/%d/%Y', birth_date), SAFE.PARSE_DATE('%Y/%m/%d', birth_date)), YEAR) AS customer_age,
      DATE_DIFF(DATE '2025-12-31', COALESCE(SAFE.PARSE_DATE('%Y-%m-%d', open_date), SAFE.PARSE_DATE('%m/%d/%Y', open_date), SAFE.PARSE_DATE('%Y/%m/%d', open_date)), MONTH) AS customer_tenure_in_months
  FROM
      dqpizza.master_customer
  WHERE
      customer_id IS NOT NULL
)
SELECT
    o.order_id,
    od.order_details_id,
    od.pizza_id,
    o.customer_id,
    o.order_maker_id,
    o.order_date,
    c.customer_name,
    c.customer_gender,
    c.customer_age,
    c.customer_tenure_in_months,
    pt.pizza_name,
    pt.pizza_category,
    p.pizza_size,
    p.pizza_price,
    p.pizza_production_cost,
    od.quantity,
    pt.pizza_ingredients,
    o.order_time,
    o.completion_time,
    o.duration_in_minutes,
    o.is_complain,
    o.complain_detail
FROM trx_order AS o
INNER JOIN trx_order_detail AS od ON o.order_id = od.order_id
INNER JOIN mst_pizza AS p ON od.pizza_id = p.pizza_id
INNER JOIN mst_pizza_type AS pt ON p.pizza_type_id = pt.pizza_type_id
INNER JOIN mst_customer AS c ON o.customer_id = c.customer_id;
```
